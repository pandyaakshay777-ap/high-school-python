- To calculate the mean, median and standard deviation for the given data <br>
  <code>[5, 6, 1, 3, 20, 18, 19, 25, 27, 15, 13, 17, 19]</code>

```python
import numpy as np

array = np.array([5, 6, 1, 3, 20, 18, 19, 25, 27, 15, 13, 17, 19])
print("Mean =", np.mean(array))
print("Median =", np.median(array))
print("Standard Deviation =", np.std(array))
```
