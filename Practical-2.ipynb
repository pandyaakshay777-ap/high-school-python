### WAP to add the elements of two lists
- Check if the lengths of both the lists are equal or not.
- Print the new list.

```python
list1 = [1, 2, 3, 4]
list2 = [10, 20, 30, 40]
```

```python
if len(list1) == len(list2):
    result = [a + b for a, b in zip(list1, list2)]
    print(result)
else:
    print("ERROR: The lengths of the lists are not equal.")
```
