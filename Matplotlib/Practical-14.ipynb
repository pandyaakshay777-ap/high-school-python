    <thead>
        <tr>
            <th>Month</th>
            <th>Sales (in ₹)</th>
        </tr>
    </thead>
    <tbody>
        <tr>
            <td>January</td>
            <td>2500</td>
        </tr>

        <tr>

    </tbody>

```python
import matplotlib.pyplot as plt
x = ["Jan", "Feb", "March", "April", "May", "June"]
y = [2500, 2100, 1700, 3500, 3000, 3800]
plt.scatter(x, y, color = "g")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.title("Sales Stats")
plt.show()
```
